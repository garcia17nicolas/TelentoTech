```python
import pandas as pd
import string
import requests
from datetime import datetime
import matplotlib.pyplot as plt

# Función para codificar con cifrado César
def codifica(s, clave):
    abc = string.ascii_lowercase
    cifrado = ''
    for c in s:
        if c in abc:
            cifrado += abc[(abc.index(c) + clave) % 26]
        else:
            cifrado += c
    return cifrado

# Función para decodificar (usando clave negativa)
def decodifica(s, clave):
    return codifica(s, -clave)

# Crear DataFrame con 10 mensajes y claves
datos = {
    'msj_original': [
        'hola',
        'hello!',
        'me gusta cripto',
        'python',
        'abc xyz',
        'seguridad',
        'dataframe',
        'clave secreta',
        'cifrado cesar',
        'ultimo mensaje'
    ],
    'clave': [1, 1, 0, 3, 4, 5, 2, 7, 8, 6]
}

df = pd.DataFrame(datos)

# Aplicar funciones de cifrado y descifrado
df['msj_cifrado'] = df.apply(lambda row: codifica(row['msj_original'], row['clave']), axis=1)
df['msj_descifrado'] = df.apply(lambda row: decodifica(row['msj_cifrado'], row['clave']), axis=1)

# Mostrar tabla resultante
print(df)


```
